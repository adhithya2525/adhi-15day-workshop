```python
import matplotlib.pyplot as plt
import pandas as pd

query6="""SELECT department,
       AVG(math_score) AS avg_math_score,
       AVG(science_score) AS avg_science_score,
       AVG(english_score) AS avg_english_score,
       AVG(programming_score) AS avg_programming_score
FROM students
GROUP BY department
ORDER BY department"""




query6_df=pd.read_sql_query(query6, conn)
query6_df = query6_df.round(2)
plot_df = query6_df.set_index('department')[['avg_math_score', 'avg_science_score', 'avg_english_score', 'avg_programming_score']]

ax = plot_df.plot(kind='bar', figsize=(10, 6))
plt.title('Average Subject Scores by Department')
plt.ylabel('Average Score')
plt.xlabel('Department')
plt.xticks(rotation=45)
plt.legend(title='Subjects')
plt.grid(axis='y', linestyle='--', alpha=0.7)

for container in ax.containers:
    ax.bar_label(container, padding=3)

plt.tight_layout()
plt.show()
```
