```python
df['quantity'].fillna('0',inplace=True)
df['customer_name'].fillna('unknown',inplace=True)
df['product'].fillna('unknown',inplace=True)
df['category'].fillna('unknown',inplace=True)
```

```python
df['order_date'] = pd.to_datetime(df['order_date'], format='mixed')
df['year']=df['order_date'].dt.year
df['month']=df['order_date'].dt.month
df['day']=df['order_date'].dt.strftime('%B')
print(df[['order_date','year','month','day']])
```

```python
df['formatted_date'] = df['order_date'].dt.strftime('%d/%m/%Y')
display(df[['order_date', 'formatted_date']].head())
```

```python
text_cols = ['customer_name', 'product', 'category', 'city', 'sales_rep']
for col in text_cols:
    if col in df.columns:
        df[col] = df[col].astype(str).str.strip().str.title()
def mask_name(name):
    parts = name.split()
    if len(parts) > 1:
        return f"{parts[0]} {parts[1][0]}***"
    return f"{parts[0][0]}***"

df['customer_name'] = df['customer_name'].apply(mask_name)

print("Text standardized and names masked.")
display(df[['customer_name', 'product', 'city', 'sales_rep']].head(10))
```

### Post-Cleaning Validation

```python
display(df.dtypes)
```

The `quantity` column is currently an `object` type because it was filled with the string `'0'`. We need to convert it to a numeric type (integer or float) for calculations. Before converting, we should ensure all values are indeed numeric or can be coerced to numeric. Let's convert it to a numeric type, coercing errors to NaN, and then fill any new NaNs with 0 (if any were created by coercion).

```python
df['quantity'] = pd.to_numeric(df['quantity'], errors='coerce')
df['quantity'].fillna(0, inplace=True)
df['quantity'] = df['quantity'].astype(int)
display(df.head())
```

Now, let's reconfirm there are no missing values and look at summary statistics.

```python
print("Missing values after 'quantity' conversion and other cleaning:\n", df.isnull().sum())
print("\nSummary statistics for numerical columns:")
display(df.describe())
print("\nSummary statistics for categorical columns:")
display(df.describe(include='object'))
```
